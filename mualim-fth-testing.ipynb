```python
import requests
import base64
import tensorflow as tf
import json

# URL endpoint model Railway
URL = "https://mualim-fth-pipeline-production.up.railway.app/v1/models/churn-model:predict"

def _bytes_feature(value):
    return tf.train.Feature(bytes_list=tf.train.BytesList(value=[value.encode('utf-8')]))
def _float_feature(value):
    return tf.train.Feature(float_list=tf.train.FloatList(value=[value]))
def _int64_feature(value):
    return tf.train.Feature(int64_list=tf.train.Int64List(value=[value]))

# bungkus data mentah menjadi format tf.train.Example
example = tf.train.Example(features=tf.train.Features(feature={
    'customerID': _bytes_feature('1234-DUMMY'), # <-- Menambahkan customerID yang diminta model
    'gender': _bytes_feature('Female'),
    'SeniorCitizen': _bytes_feature('0'),
    'Partner': _bytes_feature('Yes'),
    'StreamingTV': _bytes_feature('No'),
    'PhoneService': _bytes_feature('Yes'),
    'InternetService': _bytes_feature('DSL'),
    'PaperlessBilling': _bytes_feature('Yes'),
    'tenure': _int64_feature(1),
    'MonthlyCharges': _float_feature(29.85),
    'TotalCharges': _float_feature(29.85)
}))

# ubah ke biner dan encode menjadi base64
serialized_example = example.SerializeToString()
b64_example = base64.b64encode(serialized_example).decode('utf-8')

# masukkan kode base64 tersebut ke dalam payload JSON
data = {
    "signature_name": "serving_default",
    "instances": [
        {
            "examples": {"b64": b64_example}
        }
    ]
}

# tembak API Railway
response = requests.post(URL, json=data)

print("Status Code HTTP:", response.status_code)
print("Hasil Prediksi JSON:")
print(json.dumps(response.json(), indent=4))
```
